## 1. Variables y tipos

*Visto en 1.4 (variables), 1.6 (listas y tuplas) y 1.7 (diccionarios).*

Una **variable** es un nombre que guarda un valor. Se crea con `=` y Python deduce el tipo automáticamente. Recuerda:
- Los nombres van en minúsculas y con `_` entre palabras: `precio_cierre`.
- Podemos asignar varias variables a la vez o intercambiar sus valores en una sola línea.

````python
precio_compra = 182.5                      # Una variable
precio_apertura, precio_cierre = 101.5, 103.2  # Varias variables a la vez
comision_compra = comision_venta = 1.5     # El mismo valor para dos variables

# Intercambio de valores en una sola línea
precio_apertura, precio_cierre = precio_cierre, precio_apertura
print(precio_apertura, precio_cierre)
````

Estos son todos los tipos que hemos visto en el módulo:

|Tipo|Descripción|Ejemplo|
|-|-|-|
|`int`|Números enteros|`num_acciones = 100`|
|`float`|Números reales (también en notación científica)|`precio = 152.35`, `rentabilidad = 5.2e-2`|
|`str`|Texto|`ticker = 'AAPL'`|
|`bool`|Verdadero o falso|`mercado_abierto = True`|

|`list`|Colección ordenada que se puede modificar|`[182.5, 183.1, 180.9]`|
|`tuple`|Colección ordenada que **no** se puede modificar|`('AAPL', 182.5, 100)`|
|`dict`|Pares clave: valor|`{'AAPL': 228.88, 'MSFT': 417.32}`|

Con `type()` sabemos de qué tipo es una variable.

````python
import datetime

num_acciones = 100
rentabilidad = 5.2e-2
ticker = 'AAPL'
mercado_abierto = True
precio_cierre_hoy = None
fecha_operacion = datetime.datetime.now()
precios = [182.5, 183.1, 180.9]
posicion = ('AAPL', 182.5, 100)
cotizaciones = {'AAPL': 228.88, 'MSFT': 417.32}

print(type(num_acciones), type(rentabilidad), type(ticker))
print(type(mercado_abierto), type(precio_cierre_hoy), type(fecha_operacion))
print(type(precios), type(posicion), type(cotizaciones))
````

**[Nuevo]** Conversión entre tipos

A veces un dato nos llega con un tipo que no nos sirve. Por ejemplo, un precio leído de un fichero o pedido con `input()` es texto, y con texto no podemos hacer cuentas. Para convertirlo usamos el nombre del tipo como función (ya lo hicimos con `float(input(...))` en 1.5):

|Función|Convierte a|Ejemplo|Resultado|
|-|-|-|-|
|`int()`|Entero (quita los decimales, no redondea)|`int(9.99)`|`9`|
|`float()`|Número real|`float('182.50')`|`182.5`|
|`str()`|Texto|`str(100)`|`'100'`|
|`bool()`|Booleano (`0` y `''` son `False`, el resto `True`)|`bool(0)`|`False`|

````python
precio_texto = '182.50'               # Parece un número, pero es un str
precio = float(precio_texto)          # Ahora sí es un float
print(precio * 100)

print(int(9.99))                      # int() corta los decimales
print(str(100) + ' acciones')         # Para unir un número con texto, lo pasamos a str
print(bool(0), bool(15))
````

## 2. `print()` y strings

*Visto en 1.3, 1.4 y 1.5.*

`print()` muestra por pantalla todo lo que le pasemos separado por comas, y pone un espacio entre cada valor.

**[Nuevo]** Con `sep=` elegimos qué se pone entre los valores y con `end=` qué se pone al final (por defecto, un salto de línea).

````python
print('AAPL', 182.5, 100)
print('AAPL', 182.5, 100, sep=' | ')     # Separador personalizado
print('Descargando precios', end='... ')  # No salta de línea al terminar
print('hecho')
````

Un **string** (`str`) es un texto. Se escribe entre comillas simples, dobles o triples (estas últimas permiten varias líneas). Muchas cosas funcionan igual que en las listas:
- `+` une textos y `*` los repite.
- `len()` da el número de caracteres.
- Se indexan y se cortan igual que las listas: `texto[0]`, `texto[-1]`, `texto[:5]`.

````python
ticker = "BBVA"
nombre = 'Banco Bilbao Vizcaya Argentaria'
descripcion = """Banco español.
Cotiza en el IBEX 35."""

print(ticker + ' - ' + nombre)   # Concatenación
print('-' * 30)                  # Repetición
print(descripcion)
print(len(nombre))               # Número de caracteres
print(nombre[0], nombre[-1], nombre[:5])
````

**[Nuevo]** Métodos de los strings

Igual que las listas tienen `append()` o `sort()`, los strings tienen sus propios métodos. Estos son los más útiles para limpiar datos:

|Método|Qué hace|Ejemplo|Resultado|
|-|-|-|-|
|`<str>.upper()`|Pasa a mayúsculas|`'aapl'.upper()`|`'AAPL'`|
|`<str>.lower()`|Pasa a minúsculas|`'AAPL'.lower()`|`'aapl'`|
|`<str>.strip()`|Quita los espacios del principio y del final|`'  SAN '.strip()`|`'SAN'`|
|`<str>.replace(a, b)`|Cambia `a` por `b`|`'1,5'.replace(',', '.')`|`'1.5'`|
|`<str>.split(sep)`|Corta el texto y devuelve una lista|`'AAPL,MSFT'.split(',')`|`['AAPL', 'MSFT']`|

Estos métodos **no modifican** el string original: devuelven uno nuevo, que tenemos que guardar en una variable si lo queremos usar después.

````python
ticker_sucio = '  san  '
ticker_limpio = ticker_sucio.strip().upper()   # Se pueden encadenar
print(f"'{ticker_sucio}' -> '{ticker_limpio}'")

# Un precio en formato español: quitamos el punto de miles y cambiamos la coma por un punto
precio_es = '1.234,56'
precio = float(precio_es.replace('.', '').replace(',', '.'))
print(precio)

tickers = 'AAPL,MSFT,NVDA'.split(',')
print(tickers, type(tickers))
````

**f-strings y formato de números**

Un f-string empieza por `f` y permite meter variables (o cualquier operación) entre llaves: `f"Valor: {acciones * precio}"`.

**[Nuevo]** Después de la variable, con `:` podemos indicar cómo queremos que se muestre el número:

|Formato|Qué hace|Ejemplo con `x = 18250.456`|
|-|-|-|
|`{x:.2f}`|2 decimales|`18250.46`|
|`{x:,.2f}`|Separador de miles y 2 decimales|`18,250.46`|
|`{x:.0f}`|Sin decimales (redondea)|`18250`|
|`{r:.1%}`|Porcentaje con 1 decimal (multiplica por 100)|`r = 0.0523` → `5.2%`|
|`{r:+.2f}`|Siempre con signo|`r = 3.1` → `+3.10`|
|`{x:>12}` / `{x:<12}` / `{x:^12}`|Ocupa 12 caracteres, alineado a la derecha / izquierda / centro|`'    18250.456'`|
|`{x:>12,.2f}`|Se pueden combinar: ancho, miles y decimales|`'   18,250.46'`|

Python usa el formato inglés: coma para los miles y punto para los decimales. Para redondear un número sin mostrarlo también existe `round(x, 2)`, que ya usamos en 1.8.

````python
valor = 18250.456
rentabilidad = 0.0523
variacion = 3.1

print(f'Valor: {valor}')
print(f'Valor: {valor:.2f} €')
print(f'Valor: {valor:,.2f} €')
print(f'Rentabilidad: {rentabilidad:.1%}')
print(f'Variación: {variacion:+.2f} €')
print(round(valor, 2))
````

````python
# Con el ancho y la alineación podemos hacer tablas
print(f"{'Ticker':<6} | {'Acciones':>8} | {'Valor (€)':>12}")
print(f"{'AAPL':<6} | {100:>8} | {18250.456:>12,.2f}")
print(f"{'SAN':<6} | {1500:>8} | {6615.0:>12,.2f}")
````

## 3. Operadores

*Visto en 1.5.*

### Aritméticos

|Operador|Operación|Ejemplo|Resultado|
|-|-|-|-|
|`+` `-`|Suma y resta|`1000 + 250 - 89`|`1161`|
|`*` `/`|Multiplicación y división|`100 * 15.30 / 4`|`382.5`|
|`**`|Potencia|`1000 * 1.05 ** 3`|`1157.625`|
|`//`|División entera (sin decimales)|`1250 // 100`|`12`|
|`%`|**[Nuevo]** Resto de la división entera|`1250 % 100`|`50`|

El operador `%` (módulo) da lo que **sobra** al hacer la división entera. Va muy bien junto a `//`: si tenemos 1.250 acciones y se negocian en lotes de 100, `1250 // 100` son 12 lotes completos y `1250 % 100` son las 50 acciones sueltas. También sirve para saber si un número es par: `n % 2 == 0`.

````python
acciones = 1250
lote = 100

print('Lotes completos:', acciones // lote)
print('Acciones sueltas:', acciones % lote)
print('¿Número de acciones par?', acciones % 2 == 0)

# Comprobación: lotes * tamaño del lote + resto = total
print((acciones // lote) * lote + acciones % lote)
````

**[Nuevo]** Asignación compuesta

Cuando actualizamos una variable a partir de su propio valor, podemos escribirlo de forma abreviada. Ya apareció `anio += 1` en el bucle `while` de 1.8.

|Abreviado|Equivale a|
|-|-|
|`saldo += 250`|`saldo = saldo + 250`|
|`saldo -= 15`|`saldo = saldo - 15`|
|`saldo *= 1.05`|`saldo = saldo * 1.05`|
|`saldo /= 2`|`saldo = saldo / 2`|

````python
saldo = 1000
saldo += 250      # Aportación
saldo -= 15       # Comisión
saldo *= 1.05     # Rentabilidad del 5 %
print(f'Saldo final: {saldo:.2f} €')
````

### Comparación

Devuelven `True` o `False`. Cuidado: `=` asigna un valor y `==` compara.

|Operador|Significado|
|-|-|
|`==` / `!=`|Igual / distinto|
|`>` / `<`|Mayor / menor|
|`>=` / `<=`|Mayor o igual / menor o igual|
|`is` / `is not`|Es / no es el mismo objeto|

### Lógicos

|Operador|Devuelve `True` si…|
|-|-|
|`and`|Se cumplen **las dos** condiciones|
|`or`|Se cumple **al menos una**|
|`not`|La condición **no** se cumple|

````python
precio_actual = 12
precio_compra = 10
volumen_hoy, volumen_medio = 6, 5

print(precio_actual > precio_compra)          # ¿Estamos en ganancias?
print(precio_actual != precio_compra)         # ¿Se ha movido el precio?
print(precio_actual >= 1.2 * precio_compra)   # ¿Ha subido al menos un 20 %?

# Regla de compra: el precio sube Y el volumen supera la media
print(precio_actual > precio_compra and volumen_hoy > volumen_medio)
print(precio_actual > precio_compra or volumen_hoy > volumen_medio)
print(not (precio_actual > precio_compra))
````

## 4. Listas y tuplas

*Visto en 1.6.*

Una **lista** guarda varios elementos en orden y se puede modificar. Se indexa empezando en 0, y con índices negativos contamos desde el final. Con `[inicio:fin:paso]` cogemos un trozo (el `fin` no se incluye).

````python
precios = [182.5, 183.1, 180.9, 185.4, 186.0, 184.2, 187.7]  # Cierres de 7 sesiones

print(precios[0])      # Primer elemento
print(precios[-1])     # Último elemento
print(precios[1:4])    # Del segundo al cuarto
print(precios[:3])     # Los 3 primeros
print(precios[-3:])    # Los 3 últimos
print(precios[::2])    # Uno sí, uno no

# Listas anidadas y unpacking
datos = [['AAPL', 'MSFT'], [228.88, 417.32]]
print(datos[1][0])
apertura, maximo, cierre = [182.5, 186.0, 184.2]
print(apertura, maximo, cierre)
````

Funciones que se aplican a una lista y métodos de las listas:

|Funciones|Notas|
|---|----|
|`len(<list>)`|Número de elementos|
|`sum(<list>)`, `min(<list>)`, `max(<list>)`|Suma, mínimo y máximo|
|`sorted(<list>)`|Devuelve una copia ordenada (la original no cambia)|
|`list(range(inicio, fin, paso))`|Crea una lista de números|

|Métodos|Notas|
|---|----|
|`<list>.append(<el>)`|Añade un elemento al final|
|`<list>.extend(<list>)`|Añade los elementos de otra lista|
|`<list>.sort()` / `<list>.reverse()`|Ordena / invierte la lista (la modifica)|
|`<list>.index(<el>)`|Posición de un elemento|
|`<list>.insert(i, <el>)`|**[Nuevo]** Inserta un elemento en la posición `i`|
|`<list>.remove(<el>)`|**[Nuevo]** Elimina la primera vez que aparece el elemento|
|`<list>.pop()` / `<list>.pop(i)`|**[Nuevo]** Saca el último elemento (o el de la posición `i`) y lo devuelve|
|`<list>.count(<el>)`|**[Nuevo]** Cuántas veces aparece el elemento|

````python
dividendos = [0.24, 0.24, 0.25, 0.25]
print(len(dividendos), min(dividendos), max(dividendos), sum(dividendos))
print(sorted([186.0, 182.5, 184.2]))
print(list(range(2020, 2026)))

cartera = ['AAPL', 'MSFT', 'NVDA']
cartera.append('AMZN')
cartera.extend(['SAN', 'BBVA'])
print(cartera, cartera.index('NVDA'))
````

````python
# [Nuevo] insert, remove, pop y count
cartera = ['AAPL', 'MSFT', 'NVDA', 'MSFT']
cartera.insert(0, 'GOOGL')     # Inserta al principio
print(cartera)
cartera.remove('MSFT')         # Elimina el primer 'MSFT'
print(cartera)
ultimo = cartera.pop()         # Saca el último y lo guarda
print(ultimo, cartera)
print(cartera.count('AAPL'))
````

Con `in` y `not in` comprobamos si un elemento está en una lista. Y recuerda: `lista_b = lista_a` **no** copia la lista (las dos variables apuntan a la misma); para copiarla usamos `.copy()`.

````python
ibex = ['SAN', 'BBVA', 'ITX', 'TEF', 'IBE']
print('ITX' in ibex, 'AAPL' not in ibex)

lista_a = [100, 200, 300]
lista_b = lista_a          # Misma lista
lista_c = lista_a.copy()   # Copia independiente
lista_a[0] = 9999
print(lista_a, lista_b, lista_c)
````

Una **tupla** es como una lista pero **no se puede modificar**. Se escribe con paréntesis y es útil para datos que van juntos y no deben cambiar, como un registro `(ticker, precio, acciones)`.

````python
posicion = ('AAPL', 182.5, 100)
print(posicion[0], posicion[-1])

ticker, precio, acciones = posicion   # Unpacking
print(f'{ticker}: {acciones} acciones a {precio} €')
````

## 5. Diccionarios

*Visto en 1.7.*

Un **diccionario** guarda pares `clave: valor`. Accedemos a los valores por su clave, no por su posición. Las claves no se repiten y los valores pueden ser de cualquier tipo (números, listas, tuplas, otros diccionarios…).

````python
cotizaciones = {'AAPL': 228.88, 'MSFT': 417.32}
cotizaciones['NVDA'] = 117.99      # Añadir una clave nueva
cotizaciones['AAPL'] = 230.10      # Modificar una que ya existe
print(cotizaciones['AAPL'])

empresa = {
    'ticker': 'AAPL',
    'precios': [228.10, 229.50, 228.90],    # lista
    'rango_52_semanas': (164.08, 237.23),   # tupla
}
print(empresa['precios'][-1])
````

|`<dict>.keys()` / `.values()` / `.items()`|Claves, valores y pares clave-valor|

## 6. Condicionales

*Visto en 1.8.*

Con `if` ejecutamos un bloque solo si se cumple una condición. Con `elif` añadimos más condiciones y con `else` lo que pasa si no se cumple ninguna. Python comprueba las condiciones **en orden** y ejecuta solo el primer bloque que se cumple. El bloque se marca con la **indentación** (4 espacios).

```python
if condicion_1:
    ...
elif condicion_2:
    ...
else:
    ...
```

````python
precio_actual = 10
precio_objetivo = 12

if precio_actual > precio_objetivo:
    print('Precio por encima del objetivo: vender')
elif precio_actual == precio_objetivo:
    print('Precio en el objetivo: vender la mitad')
else:
    print('Precio por debajo del objetivo: mantener')
````

````python
# Las condiciones pueden combinar operadores lógicos
rentabilidad = 7      # en %
volatilidad = 25      # en %

if rentabilidad > 5 and volatilidad < 20:
    print('Buena rentabilidad con poco riesgo')
elif rentabilidad > 5:
    print('Buena rentabilidad, pero con mucho riesgo')
else:
    print('Rentabilidad baja')
````

## 7. Bucles `for` y `while`

*Visto en 1.8.*

- `for` recorre los elementos de una secuencia: un `range`, una lista, las claves o los pares de un diccionario…
- `enumerate()` nos da también la posición y `zip()` recorre dos listas a la vez.
- `while` repite mientras se cumpla una condición (¡hay que asegurarse de que en algún momento deja de cumplirse!).
- `break` sale del bucle y `continue` salta a la siguiente vuelta.

````python
tickers = ['AAPL', 'MSFT', 'NVDA']
precios = [228.88, 417.32, 117.99]
cartera = {'AAPL': 100, 'MSFT': 50, 'NVDA': 25}

for sesion in range(1, 4):
    print('Sesión', sesion)

for i, ticker in enumerate(tickers):
    print(i, ticker)

for ticker, precio in zip(tickers, precios):
    print(f'{ticker}: {precio:.2f} €')

for ticker, acciones in cartera.items():
    print(f'{ticker}: {acciones} acciones')
````

````python
# while: años hasta llegar a 1.500 € con un 10 % anual
capital = 1000
anio = 0
while capital < 1500:
    capital *= 1.10
    anio += 1
print(f'{anio} años, capital final: {capital:.2f} €')
````

````python
precios = [100, 102, 98, 95, 101]
for precio in precios:
    if precio < 96:       # Stop loss
        print('Stop loss en', precio)
        break             # Salimos del bucle
    print(precio)

print('---')

tickers = ['AAPL', 'MSFT', 'TSLA', 'NVDA']
for ticker in tickers:
    if ticker == 'TSLA':
        continue          # Saltamos este y seguimos con el siguiente
    print(ticker)
````

## 8. Funciones

*Visto en 1.9.*

Una función agrupa un bloque de código para reutilizarlo. Buenas prácticas:
* Nombre de la función en infinitivo.
* Indicar los tipos de los parámetros y del valor devuelto.
* Documentar la función con un docstring: qué hace, qué recibe (`Args`) y qué devuelve (`Returns`).

```python
def nombre_de_la_funcion(argumento_1: tipo, argumento_2: tipo = valor_por_defecto) -> tipo:
    """Qué hace la función.

    Args:
        argumento_1 (tipo): Descripción
        argumento_2 (tipo, opcional): Descripción. Por defecto, valor_por_defecto

    Returns:
        tipo: Descripción
    """
    Bloque de código
    return valor
```

- Los argumentos **por defecto** van siempre al final y se pueden omitir al llamar a la función.
- Al llamarla, podemos pasar los argumentos **por posición** o **por nombre** (y entonces el orden da igual).
- Si devolvemos varios valores separados por comas, la función devuelve una **tupla**, que podemos desempaquetar.

````python
def calcular_capital_final(capital: float, tipo: float = 0.03, anios: int = 5) -> float:
    """Calcula el capital final de una inversión a interés simple.

    Args:
        capital (float): Capital inicial
        tipo (float, opcional): Tipo de interés anual. Por defecto, 0.03
        anios (int, opcional): Número de años. Por defecto, 5

    Returns:
        float: Capital final
    """
    return capital * (1 + tipo * anios)


print(calcular_capital_final(1000))                   # Usa los valores por defecto
print(calcular_capital_final(1000, 0.05))             # Por posición
print(calcular_capital_final(1000, anios=10))         # Por nombre
print(calcular_capital_final(anios=2, capital=1000))  # Por nombre, el orden da igual
````

````python
def calcular_resultado(precio_compra: float, precio_venta: float, acciones: int) -> tuple:
    """Calcula el beneficio y la rentabilidad de una operación.

    Args:
        precio_compra (float): Precio de compra por acción
        precio_venta (float): Precio de venta por acción
        acciones (int): Número de acciones

    Returns:
        tuple: Beneficio en euros y rentabilidad en tanto por uno
    """
    beneficio = (precio_venta - precio_compra) * acciones
    rentabilidad = precio_venta / precio_compra - 1
    return beneficio, rentabilidad


resultado = calcular_resultado(100, 112, 50)
print(resultado, type(resultado))

beneficio, rentabilidad = calcular_resultado(100, 112, 50)   # Unpacking
print(f'Beneficio: {beneficio:.2f} €, rentabilidad: {rentabilidad:.1%}')
````
