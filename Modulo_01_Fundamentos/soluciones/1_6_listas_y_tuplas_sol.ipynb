<div style="background-color: #e0f7fa; padding: 10px;">

Supongamos que queremos preparar una orden de compra para nuestra cartera.

1. La orden tiene 2 tipos de activos: "acciones" y "ETFs".
2. Queremos comprar 3 acciones: "SAN", "BBVA" e "ITX".
3. Queremos comprar 2 ETFs: "SPY" y "GLD".
4. El precio de cada activo es el siguiente:
    * SAN: 4.5
    * BBVA: 9.8
    * ITX: 48.6
    * SPY: 560.3
    * GLD: 230.1
5. El número de títulos a comprar es:
    * SAN: 200
    * BBVA: 150
    * ITX: 20
    * SPY: 3
    * GLD: 5

Debemos calcular el importe total de la orden utilizando 3 variables:

* orden_compra: debe tener dos dimensiones (una lista de listas)
* precios
* cantidades

</div>

**Fórmulas:**

$$\text{Importe total} = \sum_{i} \text{precio}_i \times \text{cantidad}_i$$

````python
# Creamos las 3 variables
orden_compra = [["SAN", "BBVA", "ITX"], ["SPY", "GLD"]]
precios = [4.5, 9.8, 48.6, 560.3, 230.1]
cantidades = [200, 150, 20, 3, 5]

# Los precios y cantidades siguen el orden de orden_compra: primero las acciones y luego los ETFs
# Acciones (orden_compra[0]): posiciones 0, 1 y 2 de precios y cantidades
importe_acciones = precios[0] * cantidades[0] + precios[1] * cantidades[1] + precios[2] * cantidades[2]
# ETFs (orden_compra[1]): posiciones 3 y 4
importe_etfs = precios[3] * cantidades[3] + precios[4] * cantidades[4]

importe_total = importe_acciones + importe_etfs
print(f"Importe de las acciones {orden_compra[0]}: {importe_acciones:.2f} €")
print(f"Importe de los ETFs {orden_compra[1]}: {importe_etfs:.2f} €")
print(f"Importe total de la orden: {importe_total:.2f} €")
````

## Funciones útiles con listas

* La función **range()** sirve para generar listas de números enteros.
* No se crea la lista de números hasta que no se usa.

Desde 0 hasta n-1 --> range(n)

````python
range(10) # Indica el rango, pero no lo crea en memoria
````

````python
list(range(10)) # Si lo convertimos a una lista, se crea en memoria (p. ej. los 10 primeros días de un periodo)
````

Desde m hasta n-1 --> range(m, n)

````python
list(range(2015, 2025))  # Años de un histórico de cotizaciones
````

Desde m hasta n-1 de s en s --> range(m, n, s)

````python
list(range(0, 100, 10))  # Porcentajes de 10 en 10 para un análisis de sensibilidad
````

<div style="background-color: #e0f7fa; padding: 10px;">

Tenemos dos listas con los tickers de algunas acciones y queremos hacer varias cosas:
- Buscar en la web de yahoo finance el ticker de 2 acciones nuevas y añadir una a cada lista.
- Juntar las dos listas en una nueva.
- Ordenar la nueva lista por orden alfabético.
- Ahora de manera inversa.
- Intercambiar los valores entre el primero y el último.
- Crear una copia de la lista y ordenarla por longitud.

````python
tickers_1 = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'TSLA']
tickers_2 = ['META', 'NFLX', 'NVDA', 'JPM', 'V']

# Escribe aquí tu código

# Añadimos los tickers a las listas (Coca-Cola y Disney)
tickers_1.append('KO')
tickers_2.append('DIS')

# Juntamos las listas
tickers_total = tickers_1 + tickers_2
print("Listas juntas:      ", tickers_total)

# Ordenamos la lista
tickers_total.sort()
print("Orden alfabético:   ", tickers_total)

# Ordenar inversa
tickers_total.sort(reverse=True)
print("Orden inverso:      ", tickers_total)

# Intercambiamos el primero con el último
tickers_total[0], tickers_total[-1] = tickers_total[-1], tickers_total[0]
print("Primero <-> último: ", tickers_total)

# Creamos una copia y la ordenamos por longitud
tickers_por_longitud = tickers_total.copy()
tickers_por_longitud.sort(key=len)
print("Por longitud:       ", tickers_por_longitud)
````
