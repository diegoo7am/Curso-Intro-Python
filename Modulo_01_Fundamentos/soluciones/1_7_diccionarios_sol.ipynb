# Diccionarios

- Los diccionarios son conjuntos de datos `clave : valor`

````python
d = dict()
d = {}
print(type(d))
````

````python
# Para añadir elementos al diccionario
d['AAPL'] = 228.88
d['precios_MSFT'] = [415.1, 417.3, 420.0]
print(d)
````

````python
# Para recuperar los valores...
d['AAPL']
````

````python
# Tambien podemos definir el diccionario de manera manual
d = {
    'AAPL' : 228.88,
    'MSFT' : 417.32,
    'NVDA' : 117.99
}
````

````python
d['AAPL']
````

- Las claves también pueden ser números, pero siguen siendo claves, no indexables

````python
# Dividendo anual por acción
d = {
    2022: 0.92,
    2023: 0.96,
    2024: 1.00
}
````

````python
d[2023]
````

````python
# Causa error
d[2022:2023]
````

- Los valores asociados a cada clave pueden ser cualquier objeto

````python
d = {
    'precios': [228.10, 229.50, 228.90, 230.40],             # lista
    'rango_52_semanas': (164.08, 237.23),                    # tupla
    'balance': [[365.0, 308.0], [57.0, 42.0]],               # matriz (activo/pasivo de 2 años)
    'ratios': {
        'PER': 34.5
    }

}
````

````python
print(d['precios'])
print(d['rango_52_semanas'])
print(d['balance'])
print(d['ratios'])
````

## Built-in Functions

|Métodos|Description|
|----|---|
```<view> = <dict>.keys()``` |                          Obtenemos las claves (keys) del diccionario.|
```<view> = <dict>.values()``` |                        Obtenemos los valores del diccionario.|
```<view> = <dict>.items()```   |                       Obtenemos los pares de clave:valor del diccionario.|
```<dict> = dict(zip(keys, values)) ```  |              Crea un diccionario desde dos colecciones.|

````python
# Creamos un diccionario
d = {'AAPL': 228.88, 'AMZN': 187.93, 'TEF': 4.41}
````

````python
# Obtenemos sus claves por separado
d.keys()
````

````python
# Obtenemos sus valores por separado
d.values()
````

````python
# Obtenemos ambos a la vez
d.items()
````

````python
# Creamos un diccionario usando la función zip() y convirtiéndolo a diccionario
d = dict(zip(['AAPL', 'AMZN', 'TEF'], [228.88, 187.93, 4.41]))
d
````

````python
# Obtenemos la longitud de la lista de claves del diccionario
print(len(d))
````

````python
# Usamos el operador 'in' para saber si una clave existe en la lista  de claves del diccionario
'AAPL' in d
````

## Práctica

<div style="background-color: #e0f7fa; padding: 10px;">

Tenemos los siguientes tickers de acciones: AAPL, AMZN, TEF, NVDA, DIS:
- Consulta en la web de yahoo finance sus valores.
- Crea un diccionario asignando Ticker -> Valor.
- Imprime el contenido del diccionario.

````python
# Escribe aquí tu código

# Añadimos los valores manualmente
ticker_valor = {
    "AAPL": 228.88,
    "AMZN": 187.93,
    "TEF": 4.41,
    "NVDA": 117.99,
    "DIS": 97.31
}

print(f"Diccionario entero: {ticker_valor}\n")
print(f"Claves: {ticker_valor.keys()}\n")
print(f"Valores: {ticker_valor.values()}\n")
print(f"Pares clave-valor: {ticker_valor.items()}")
````

<div style="background-color: #e0f7fa; padding: 10px;">


Repite el ejercicio de arriba pero accediendo a los valores de las listas.

Asigna el primero de una con el primero de la otra y viceversa.

Imprime el resultado por pantalla.

````python
valores = [228.88, 187.93, 4.41, 117.99, 97.31]
tickers = ['AAPL', 'AMZN', 'TEF', 'NVDA', 'DIS']

# Escribe aquí tu código

# Accediendo por posición a cada lista
diccionario_tickers_manual = {
    tickers[0]: valores[0],
    tickers[1]: valores[1],
    tickers[2]: valores[2],
    tickers[3]: valores[3],
    tickers[4]: valores[4]
}
print(f"Diccionario manual: {diccionario_tickers_manual}\n")

# La función zip() asocia los elementos de las listas y dict() crea el diccionario
diccionario_tickers = dict(zip(tickers, valores))
print(f"Diccionario con dict(zip()): {diccionario_tickers}")
````

<div style="background-color: #e0f7fa; padding: 10px;">

Tenemos una cartera con 120 acciones de cada uno de los tickers. Calcula el valor total de la cartera

**Fórmulas:**

$$\text{Valor de la cartera} = \sum_{i} 120 \times \text{precio}_i$$

````python
# Escribe aquí tu código
num_acciones = 120

# Clave a clave: valor de cada posición = precio × nº de acciones
valor_cartera = (diccionario_tickers['AAPL'] * num_acciones
                 + diccionario_tickers['AMZN'] * num_acciones
                 + diccionario_tickers['TEF'] * num_acciones
                 + diccionario_tickers['NVDA'] * num_acciones
                 + diccionario_tickers['DIS'] * num_acciones)
print(f"Valor total de la cartera (clave a clave): {valor_cartera:.2f} €")

# Otra forma: sum() suma todos los precios del diccionario y luego multiplicamos por el nº de acciones
valor_cartera = sum(diccionario_tickers.values()) * num_acciones
print(f"Valor total de la cartera (con sum): {valor_cartera:.2f} €")
````
