# Funciones

Una función en programación es similar a una receta de cocina. Así como una receta te dice los pasos para hacer un plato específico, una función te dice los pasos para realizar una tarea específica.

En Python, una función se define con la palabra clave `def` y se le da un nombre. La función puede recibir **parámetros**, que son como los ingredientes en una receta, y devuelve un **resultado**, que sería el plato terminado. Aquí tienes un ejemplo sencillo:

```python
def calcular_valor(acciones, precio):
    resultado = acciones * precio
    return resultado
```

En este caso, `calcular_valor` es el nombre de la función, y `acciones` y `precio` son los parámetros. La función toma estos dos números, los multiplica y luego devuelve el resultado. Puedes "llamar" a esta función con diferentes números para obtener el valor de distintas posiciones, así:

```python
print(calcular_valor(10, 25))  # Esto imprimirá 250, el valor de 10 acciones a 25 €.
```

Las funciones son útiles porque te permiten reutilizar código sin tener que escribirlo nuevamente, y hacen que tu programa sea más organizado y fácil de entender.

Además, cuando defines una función, estás configurándola con instrucciones específicas. Estas instrucciones se ejecutarán cada vez que llames a la función, **sin importar dónde esté en el código**.

Las instrucciones básicas para definir una función son:

``` python
def nombre_de_la_funcion(argumento_1,argumento_2,...,argumento_n):
    '''Documentación'''
    Bloque de codigo
    return <Valor>
```

Buenas prácticas a la hora de definir una función:
* Nombre de la función en infinitivo.
* Indicar los tipos de los parámetros recibidos.
* Indicar el tipo del valor devuelto.
* Documentar la función.

## Definición

````python
def calcular_valor_posicion(num_acciones: int, precio: float) -> float:
    """Calcula el valor de una posición.

    Args:
        num_acciones (int): Número de acciones
        precio (float): Precio de cada acción

    Returns:
        float: Valor de la posición
    """
    resultado = num_acciones * precio
    return resultado
````

````python
valor = calcular_valor_posicion(100, 182.5)
print(valor)
````

¿Cómo se llama a las funciones?

- Si no se ponen los nombres de los argumentos, toman valor en el orden en el que se pasan.
- Los argumentos pueden ser pasados en cualquier orden si se ponen los nombres.
- Si se pone nombre a un argumento, a partir de ahí se tienen que poner los nombres siempre.

````python
def calcular_patrimonio(acciones: float, bonos: float, liquidez: float) -> float:
    """Suma el valor de las tres partes de una cartera.

    Args:
        acciones (float): Valor invertido en acciones
        bonos (float): Valor invertido en bonos
        liquidez (float): Dinero en efectivo

    Returns:
        float: Patrimonio total
    """
    resultado = acciones + bonos + liquidez
    return resultado
````

````python
calcular_patrimonio(10000, 5000, 2000)
````

````python
calcular_patrimonio(acciones=10000, bonos=5000, liquidez=2000)
````

````python
calcular_patrimonio(10000, liquidez=2000, bonos=5000)
````

## Argumentos por defecto

En el siguiente ejemplo calculamos el capital final de una inversión a **interés simple**:

$$\text{Capital final} = C \times (1 + r \times t)$$

donde $C$ es el capital inicial, $r$ el tipo de interés anual y $t$ el número de años. Si al llamar a la función no indicamos $r$ o $t$, se usan los valores por defecto.

````python
def calcular_capital_final(capital: float, tipo: float = 0.03, anios: int = 5) -> float:
    """Calcula el capital final de una inversión a interés simple.

    Args:
        capital (float): Capital inicial
        tipo (float, optional): Tipo de interés anual. Defaults to 0.03.
        anios (int, optional): Número de años. Defaults to 5.

    Returns:
        float: Capital final = capital * (1 + tipo * anios)
    """
    capital_final = capital * (1 + tipo * anios)
    return capital_final
````

````python
calcular_capital_final(1000)
````

````python
calcular_capital_final(capital=1000, tipo=0.05, anios=2)
````

````python
def duplicar_capital(capital):
    print(f"Valor de capital dentro de la función antes de modificarlo = {capital}")

    capital = capital * 2

    print(f"Valor de capital dentro de la función después de modificarlo = {capital}")

    return capital

capital = 3000
print(f"capital antes de llamar a la función = {capital}")

duplicar_capital(capital)

print(f"capital después de llamar a la función = {capital}")
````

````python
comision = 10

def invertir(capital):
    global comision # Indicamos que queremos usar dentro de "invertir" la variable "comision" que está declarada fuera de ella

    print(f"Valor de capital dentro de la función antes de modificarlo = {capital}")
    print(f"Valor de comision dentro de la función antes de modificarlo = {comision}")

    capital = capital - comision
    comision = 5

    print(f"Valor de capital dentro de la función después de modificarlo = {capital}")
    print(f"Valor de comision dentro de la función después de modificarlo = {comision}")

    return capital

capital = 3000
print(f"capital antes de llamar a la función = {capital}")
print(f"comision antes de llamar a la función = {comision}")

invertir(capital)

print(f"capital después de llamar a la función = {capital}")
print(f"comision después de llamar a la función = {comision}")
````

## Valores de retorno de una función

La función del ejemplo devuelve dos resultados de una operación de compraventa:

$$\text{Beneficio} = \text{precio de venta} - \text{precio de compra} \qquad \text{Rentabilidad} = \frac{\text{Beneficio}}{\text{precio de compra}}$$

````python
def calcular_resultado(precio_compra, precio_venta):
    beneficio = precio_venta - precio_compra
    rentabilidad = beneficio / precio_compra
    return rentabilidad, beneficio
````

````python
# Podemos devolver una tupla.
resultado = calcular_resultado(100, 112)
print(resultado)
print(type(resultado))
````

````python
# O podemos hacer unpacking.
rentabilidad, beneficio = calcular_resultado(100, 112)
print(rentabilidad)
print(beneficio)
````
