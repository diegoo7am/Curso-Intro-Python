# Condicionales

Las instrucciones condicionales se utilizan para **ejecutar o no** determinadas **partes del código** en función de **si se cumplen o no** algunas **condiciones**.

Al final, son como las decisiones que tomamos al invertir. Imagina que tienes una decisión que tomar: si el precio de una acción cae por debajo de tu precio objetivo, la compras. Si no, esperas.

```
si el precio está por debajo del objetivo entonces
    comprar
si no
    esperar
```

## if
```python
if condicion:
    codigo a ejecutar
```

- La condición puede ser `True` o `False`, se usan operadores lógicos (`and` / `or`) o relacionales (`>, <, >=, <=`)
- La línea tiene que terminar en `:`
- El bloque de código se ejecuta cuando **se cumple** la condición del `if`

````python
precio_bajo_objetivo = True

if precio_bajo_objetivo == True: # Se lee como "Si el precio está por debajo del objetivo entonces..."
    comprar = True
````

Aquí, `precio_bajo_objetivo` es la condición que estamos comprobando. Si `precio_bajo_objetivo` es `True` (es decir, si el precio está por debajo de nuestro objetivo), entonces el código dentro del bloque `if` se ejecutará, y `comprar` se establecerá en `True`.

````python
rentabilidad = 7  # en %

if rentabilidad > 5: # Se lee como "si la rentabilidad es mayor que 5 entonces..."
    print("La rentabilidad es mayor que el 5 %")
````

## if else
```python
if condicion:
    codigo a ejecutar
else:
    otro codigo a ejecutar
```
- Se añade otra línea con `else` para contemplar todas las situacion que no se han tenido en cuenta en el caso del `if`
- También tiene su propio codigo que solo se ejecuta cuando **no se cumple** la condición del `if`

````python
precio_bajo_objetivo = False

if precio_bajo_objetivo == True:  # Se lee como "Si el precio está por debajo del objetivo entonces..."
    comprar = True
else:
    comprar = False # Se lee como "Si no..." 
````

````python
comprar
````

Aquí, `precio_bajo_objetivo` es la condición que estamos comprobando. Si `precio_bajo_objetivo` es `True` (es decir, si el precio está por debajo del objetivo), entonces el código dentro del bloque `if` se ejecutará, y `comprar` se establecerá en `True`.

Si `precio_bajo_objetivo` es `False` (el precio está por encima del objetivo), entonces el código dentro del bloque `else` se ejecutará, y `comprar` se establecerá en `False`.

````python
rentabilidad = 7  # en %

if rentabilidad > 5: # Se lee como "si la rentabilidad es mayor que 5 entonces..."
    print("La rentabilidad es mayor que el 5 %")
else: # Se lee como "si no..."
    print("La rentabilidad es menor o igual que el 5 %")
````

````python
rentabilidad = 4  # en %

if rentabilidad > 5:
    print("La rentabilidad es mayor que el 5 %")
else:
    print("La rentabilidad es menor o igual que el 5 %")
````

## elif
```python
if condicion:
    codigo a ejecutar 1
elif otra condicion:
    codigo a ejecutar 2
else:
    otro codigo a ejecutar 3
```

````python
precio_actual = 10
precio_objetivo = 12

if precio_actual > precio_objetivo: # Se lee como "si el precio es mayor que el objetivo entonces..."
    print("Precio por encima del objetivo: vender")
elif precio_actual < precio_objetivo: # Se lee como "si no, si el precio es menor que el objetivo entonces..."
    print("Precio por debajo del objetivo: comprar")
else: # Se lee como "si no..."
    print("Precio igual al objetivo: mantener")
````

# Bucles

Los bucles se utilizan para **repetir** un bloque de instrucciones **hasta que o mientras que** se cumpla una **condición**.

## for
```python
for variable in algo:
    codigo a ejecutar
```

- Ese algo puede ser rango, lista, tupla....

````python
# Podemos usar un range para iterar sobre una secuencia de números (p. ej. las 5 sesiones de la semana)
for i in range(5): # Se lee como "para cada i en el rango de 0 a 5..."
    print(i)
````

````python
# podemos iterar una lista
tickers = ['AAPL', 'MSFT', 'NVDA']
for i in tickers: # Se lee como "para cada i en tickers..."
    print(i)
````

- Los diccionarios se pueden iterar de diversas formas

````python
# podemos iterar un diccionario con sus claves
cartera = {'AAPL': 100, 'MSFT': 50, 'NVDA': 25}  # ticker: nº de acciones
for i in cartera.keys(): # Se lee como "para cada i en las claves de cartera..."
    print(i)
````

````python
# podemos iterar un diccionario con sus valores
cartera = {'AAPL': 100, 'MSFT': 50, 'NVDA': 25}
for i in cartera.values(): # Se lee como "para cada i en los valores de cartera..."
    print(i)
````

````python
# podemos iterar un diccionario con sus valores y claves y desempaquetarlos
cartera = {'AAPL': 100, 'MSFT': 50, 'NVDA': 25}
for i, j in cartera.items(): # Se lee como "para cada i, j en los items de cartera..."
    print(i, j)
````

- Se pueden programar bucles anidados

````python
# Una lista que contiene, a su vez, otras listas: precios de 3 activos durante 3 sesiones.
precios_activos = [[182.5, 183.1, 180.9], [415.1, 417.3, 420.0], [117.9, 119.2, 118.4]]

for i in precios_activos: # Se lee como "para cada i en precios_activos..."
    print(i)
    for j in i: # Se lee como "para cada j en i..."
        print(j)
````

- Se pueden usar muchas funciones para ayudar en la iteración:
    - enumerate()
    - zip()
    - sorted()
    - reversed()

````python
for anio in reversed([2022, 2023, 2024]):
    print(anio)
````

````python
tickers = ['AAPL', 'MSFT', 'NVDA']

for i, ticker in enumerate(tickers): # con enumerate tenemos el índice y el valor
    print(i, ticker)
````

````python
tickers = ['AAPL', 'MSFT', 'NVDA']
precios = [228.88, 417.32, 117.99]

for i, j in zip(tickers, precios): # zip nos permite iterar sobre dos listas al mismo tiempo
    print(i, j)
````

## while
```python
while condicion:
    codigo a ajecutar
```

- Con while mientras que se cumpla la condicion se ejecuta el código en bucle

````python
capital = 1000
anio = 0

while capital < 1500: # Se lee como "mientras el capital sea menor que 1500..."
    capital = capital * 1.10  # Rentabilidad del 10 % anual
    anio += 1
    print(anio, round(capital, 2))

print("fin")
````

## break

- Se utiliza para acabar un bucle

````python
precios = [100, 102, 98, 95, 101]

for precio in precios:
    print(precio)
    if precio < 96:  # Stop loss
        print("Precio por debajo del stop loss: vendemos y salimos del bucle")
        break
````

## continue

- Se utiliza para saltar una iteracion, continua con el bucle

````python
tickers = ['AAPL', 'MSFT', 'TSLA', 'NVDA']

for ticker in tickers:
    if ticker == 'TSLA':  # Excluimos este activo del análisis
        continue
    print(ticker)
````
