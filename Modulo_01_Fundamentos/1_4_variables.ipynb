# Variables

Antes de ver lo que es una variable, veamos muy por encima cómo es un ordenador por dentro y cuáles son sus componentes principales.

<img src="https://raw.githubusercontent.com/diegoo7am/Curso-Intro-Python/main/Modulo_01_Fundamentos/imgs/estructura_pc.png" alt="Estructura de un ordenador" width="75%">

Una variable se puede interpretar como una caja donde guardo un determinado valor en un momento del tiempo. Se denomina ```variable``` porque su valor puede ```variar``` a lo largo del tiempo (como la cotización de una acción); en caso contrario se denominaría ```constante```.

Cuando tengo muchas variables, necesito disponer de una manera de identificarlas para poder referirme a ellas, ver qué contenido tienen o cambiarlo o combinarlo u operar con él. Es decir, necesito una manera de ```nombrar``` variables.

<img src="https://raw.githubusercontent.com/diegoo7am/Curso-Intro-Python/main/Modulo_01_Fundamentos/imgs/variables_1.png" alt="Variables">

Existen ciertas reglas para nombrar las variables en Python:
- Caracteres alfanuméricos: a-z,A-Z,0-9
- Guión bajo: _
- No pueden empezar por un número
- No se deben usar palabras reservadas:
    - False, None, True, and, as, assert, break, class, continue, def, del, elif, else, except, finally, for, from, global, if, import, in, is, lambda, nonlocal, not, or, pass, raise, return, try, while, with, yield

La asignación de una variables es el momento en el cual le proporcionamos un valor, por ejemplo:

````python
precio_accion = 8
precio_accion
````

````python
num_acciones = 9
print(num_acciones)
````

- Como podemos ver la asignación se realiza con el operador "="
- Se pueden hacer asignaciones multiples en una sola línea

````python
precio_apertura, precio_cierre = 101.5, 103.2
print(precio_apertura)
print(precio_cierre)
````

````python
# tambien podemos asignar a multiples variables un mismo valor
comision_compra = comision_venta = 1.5
print(comision_compra)
print(comision_venta)
````

Gracias a la asignación multiple en la misma línea, podemos hacer intercambios de variables en una sola línea. Por ejemplo, si hemos introducido los precios de apertura y cierre al revés:

````python
# Manera tradicional
aux = precio_apertura
precio_apertura = precio_cierre
precio_cierre = aux
print(precio_apertura, precio_cierre)
````

````python
# Python
precio_apertura, precio_cierre = precio_cierre, precio_apertura
print(precio_apertura, precio_cierre)
````

## Tipos básicos:

|Tipo| Descripción|
|-|-|
|None|El es valor nulo en Python|
|str|Tipo cadena de caracteres, guarda caracteres de tipo UTF-8|
|float|Números reales (2.3), pudiendo estar representados en coma flotante (2.3e0)|
|bool|True o False, es el tipo usado para representar variables lógicas|
|int|Números enteros|
|datetime| Fechas|

Vamos a mostrar como se representan en python estos tipos

````python
num_acciones = 100 # int
print(num_acciones)

variacion_acciones = -20 # int (hemos vendido 20 acciones)
print(variacion_acciones)
````

````python
precio, rentabilidad = 152.35, 5.2e-2 # float
print(precio, rentabilidad)
````

````python
x = "100" # str (parece un número, pero es texto)
print(x)

x = "Apple Inc." # str
print(x)

x = 'Apple Inc.' # str
print(x)

x = """Apple Inc.
Sector: Tecnología
Mercado: NASDAQ""" # str
print(x)

x = "El ticker de Apple es 'AAPL'" # str
print(x)

x = "Ticker:\tAAPL \nPrecio:\t152.35" # str
print(x)

x = "Ruta del fichero: C:\\datos\\precios.csv" # str
print(x)

precio = 152.35
x = f"El precio de AAPL es {precio} $" # str
print(x)
````

````python
mercado_abierto = True # bool
print(mercado_abierto)

mercado_abierto = False # bool
print(mercado_abierto)
````

Para el manejo de fechas en Python tenemos varias librerías a nuestra disposición. La básica es `datetime`, que podremos utilizar haciendo uso de la instrucción `import datetime`

````python
import datetime

fecha_operacion = datetime.datetime.now() # datetime
print(fecha_operacion)
````

````python
precio_cierre_hoy = None # None Type (todavía no se conoce)
print(precio_cierre_hoy)
````
