**Fórmulas:**

$$BAI = \text{Ingresos} - \text{Aprovisionamiento} - \text{Amortización} - \text{Costes generales}$$

**¿Qué es la amortización?** Es el gasto contable que refleja el desgaste de los activos (maquinaria, instalaciones...). Reduce el beneficio, pero no supone una salida de dinero en ese año.

**EJEMPLO RESUELTO** Los precios de los tipos de bolígrafos que vendo están recogidos en la lista `B`, y las cantidades vendidas de cada uno de ellos en la lista `Q`. Determina el volumen de ventas.

**Fórmulas:**

$$\text{Ventas} = \sum_{i} \text{precio}_i \times \text{cantidad}_i$$

Esta suma de productos es justo lo que calcula el producto escalar `Q @ B`.

```python
B=[3, 12.5, 28]
Q=[200, 350, 75]
```

```python
# Convertimos las listas en arrays de NumPy para poder operar con ellas
B = np.array(B)
Q = np.array(Q)

# @ es el producto de matrices: multiplica cada cantidad por su precio y suma todo
ventas = Q @ B          # equivalente a np.dot(Q, B)
print(f"Volumen de ventas: {ventas:,.2f} €")   # :,.2f -> separador de miles y 2 decimales
```

**EJEMPLO RESUELTO** Si la inversión es de 100000 y la rentabilidad exigida del 7% convendria realizar la inversion.
$$VAN=I_0+\sum_{t=1}^T \frac{FC_t}{(1+k)^t}$$

$I_0$ es la inversión inicial y va con signo **negativo** porque es un pago. $FC_t$ es el flujo de caja del año $t$ y $k$ la rentabilidad exigida. Si $VAN > 0$, la inversión crea valor y conviene realizarla.

```python
# Usamos los flujos de caja (FC) calculados anteriormente
I0 = -100000                     # la inversión inicial es un pago, por eso es negativa
k = 0.07                         # rentabilidad exigida
t = np.arange(1, len(FC) + 1)    # años 1, 2, ..., 5

# Elemento a elemento: cada flujo se divide entre (1 + k) elevado a su año.
# np.sum suma todos esos valores actuales
VAN = I0 + np.sum(FC / (1 + k) ** t)
print(f"VAN = {VAN:,.2f} €")

# Si el VAN es positivo, la inversión crea valor
if VAN > 0:
    print("Conviene realizar la inversión")
else:
    print("No conviene realizar la inversión")
```

## Determinante de una matriz
El determinante es un número asociado a una matriz cuadrada que indica, entre otras cosas, si la matriz puede invertirse.

La idea clave es:

Si $\det(A)\neq 0$ → la matriz tiene inversa y el sistema \(Ax=b\) tiene solución única.

Si $\det(A)=0$ → la matriz no es invertible.

```python
A = np.array([
    [2, 1, 0],
    [1, 3, 1],
    [0, 2, 4]
])
```

```python
np.linalg.det(A)
```

NumPy “expande” virtualmente el array más pequeño para poder hacer la operación, sin necesidad de copiarlo manualmente.

## Regla para memorizar

NumPy compara las dimensiones desde la derecha. Dos dimensiones son compatibles si son iguales o si una de ellas vale 1.

(3,5)
(5,)

(1,5)

No compatible:

# Introducimos la incertidumbre: **np.random**

- los costes,
- los tipos de interés,

- los precios de los activos.

`np.random` permite generar valores aleatorios para representar esa incertidumbre y construir distintos escenarios.

En una simulación tenemos una matriz con una fila por escenario y una columna por año. Para obtener un resultado por escenario hay que sumar cada fila.

- `axis=1`: suma cada fila y da un resultado por fila.
- `axis=0`: suma cada columna y da un resultado por columna.

```python
ejemplo = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(ejemplo.shape)         # (2, 3): 2 filas y 3 columnas
print(ejemplo.sum(axis=1))   # una suma por fila: 10+20+30 y 40+50+60
print(ejemplo.sum(axis=0))   # una suma por columna: 10+40, 20+50 y 30+60
```

**EJEMPLO RESUELTO** Tenemos que determinar cuanto pagaríamos por una empresa, que no tiene deuda. El coste de capital es del 7%. Para ello vamos a generar 10.000 posibles escenarios para posteriormente estimar el valor de la empresa como una variable aleatoria.

Hipótesis:
- Los ingresos de los próximos 5 años son los simulados en `Ingresos_sim` (10.000 escenarios × 5 años).
- El flujo de caja libre de cada año es el 20 % de los ingresos.
- A partir del año 5 el flujo de caja crece un 2 % anual a perpetuidad: $VT_5=\frac{FC_5\times(1+g)}{k-g}$

$$Valor=\sum_{t=1}^{5}\frac{FC_t}{(1+k)^t}+\frac{VT_5}{(1+k)^5}$$

**¿Qué es el valor terminal?** Es el valor, en el año 5, de todos los flujos de caja posteriores, suponiendo que crecen a un ritmo $g$ para siempre. Por eso hay que descontarlo también al presente.

```python
k = 0.07   # coste de capital
g = 0.02   # crecimiento a perpetuidad
t = np.arange(1, 6)   # años 1 a 5

FC_sim = Ingresos_sim * 0.20                              # (10000, 5): un flujo por escenario y año
# Broadcasting: (10000, 5) / (5,). axis=1 suma por filas: un valor actual por escenario
VA_FC = (FC_sim / (1 + k) ** t).sum(axis=1)               # valor actual de los 5 flujos
VT = FC_sim[:, -1] * (1 + g) / (k - g)                    # valor terminal con el flujo del año 5 ([:, -1] = última columna)
valor_empresa = VA_FC + VT / (1 + k) ** 5                 # un valor por escenario

valor_empresa.shape
```

```python
print(f"Valor medio de la empresa: {valor_empresa.mean():,.0f} €")
print(f"Desviación típica:         {valor_empresa.std():,.0f} €")
# Percentil 5 %: solo en el 5 % de los escenarios el valor es menor que esta cifra
print(f"Percentil 5 %:             {np.percentile(valor_empresa, 5):,.0f} €")
print(f"Percentil 95 %:            {np.percentile(valor_empresa, 95):,.0f} €")

plt.figure(figsize=(10, 4))
plt.hist(valor_empresa, bins=50)
plt.axvline(valor_empresa.mean(), color="black", linestyle="--", label="Media")   # línea vertical en la media
plt.xlabel("Valor de la empresa (€)")
plt.ylabel("Frecuencia")
plt.title("Distribución del valor de la empresa")
plt.legend()
plt.grid(True)
plt.show()
```

**EJEMPLO RESUELTO - Pregunta 4** Si la desviación típica de los ingresos representa el 15% de la media y la desviación típica del aprovisionamiento 8% de la media. Realiza un análisis de Monte Carlo con 10.000 simulaciones y analiza los resultados.

**Fórmulas:**

$$\text{Ingresos} = \mu_I + \sigma_I \times z, \quad \sigma_I = 0.15 \times \mu_I \qquad\qquad \text{Aprov.} = \mu_A + \sigma_A \times z, \quad \sigma_A = 0.08 \times \mu_A$$

con $z \sim N(0,1)$ distinto para cada variable. Para cada simulación se calcula el VAN y, al final:

$$P(VAN > 0) = \frac{\text{nº de simulaciones con } VAN > 0}{\text{nº total de simulaciones}}$$

```python
np.random.seed(42)   # fija los números aleatorios para que el resultado sea siempre el mismo
n_simulaciones = 10000

# Shocks independientes N(0,1) para ingresos y aprovisionamientos: una fila por simulación y una columna por año
z_ingresos = np.random.normal(0, 1, size=(n_simulaciones, 5))
z_aprov = np.random.normal(0, 1, size=(n_simulaciones, 5))

# Media + desviación típica × shock (15 % de la media para ingresos y 8 % para aprovisionamientos)
ingresos_sim = ingresos + ingresos * 0.15 * z_ingresos
aprov_sim = aprovisionamientos + aprovisionamientos * 0.08 * z_aprov

# Flujo de caja y VAN de cada simulación (axis=1: un VAN por fila)
FC_sim = (ingresos_sim - aprov_sim - amortizacion - costes_generales) * (1 - tipo_impositivo) + amortizacion
VAN_sim = -I0 + (FC_sim * factores_descuento).sum(axis=1)

print(f"VAN medio:            {VAN_sim.mean():,.0f} €")
print(f"Desviación típica:    {VAN_sim.std():,.0f} €")
# VAN_sim > 0 da True/False en cada simulación. True cuenta como 1 y False como 0,
# así que la media de esos valores es la proporción de simulaciones con VAN positivo
print(f"Probabilidad VAN > 0: {np.mean(VAN_sim > 0):.2%}")
# Percentil 5 %: solo en el 5 % de las simulaciones el VAN es menor que esta cifra
print(f"Percentil 5 %:        {np.percentile(VAN_sim, 5):,.0f} €")
print(f"Percentil 95 %:       {np.percentile(VAN_sim, 95):,.0f} €")

plt.figure(figsize=(10, 4))
plt.hist(VAN_sim, bins=50)
plt.axvline(0, color="red", linestyle="--", label="VAN = 0")
plt.xlabel("VAN (€)")
plt.ylabel("Frecuencia")
plt.title("Distribución del VAN (10.000 simulaciones)")
plt.legend()
plt.grid(True)
plt.show()
```

**EJEMPLO RESUELTO** Suponemos que los ingresos y los costes tienen un coeficiente de correlación  $\rho=0.70$

```python
ingresos = np.array([
    500000,
    550000,
    600000,
    650000,
    700000
])

aprovisionamientos = np.array([
    180000,
    195000,
    210000,
    225000,
    240000
])

desv_ingresos = ingresos * 0.15
desv_aprovisionamientos = aprovisionamientos * 0.08

rho = 0.70

n_simulaciones = 10000
years = np.arange(1, 6)   # 5 años de vida del proyecto
```

Vamos a construir dos matrices aleatorias $N(0,1)$ independientes: $Z_1$ y $Z_2$

```python
z1 = np.random.normal(
    0,
    1,
    size=(n_simulaciones, len(years))
)

z2 = np.random.normal(
    0,
    1,
    size=(n_simulaciones, len(years))
)
```

Introducimos la correlación la función de distribución estandarizada del aprovisionamiento será:

$$z_A=\rho Z_I+\sqrt{1-\rho^2}Z_2$$

donde:
- $Z_I\sim N(0,1)$ shock de los ingresos
- $Z_2\sim N(0,1)$ shock independiente del aprovisionamiento
- $\rho$ correlación
- $z_A$ el shock del aprovisionamiento

```python
z_A = (
    rho * z1
    + np.sqrt(1 - rho**2) * z2
)
```

Transformamos el shock $z_I$ en ingresos

$$I=\mu_I + \sigma z_I$$

```python
ingresos_sim = (
    ingresos
    + desv_ingresos * z1
)
ingresos_sim
```

Transformamos el shock $z_A$ en costes de aprovisionamiento

$$Aprov=\mu_A + \sigma z_A$$

```python
aprovisionamientos_sim = (
    aprovisionamientos
    + desv_aprovisionamientos * z_A
)
aprovisionamientos_sim
```

```python
# Comprobamos si están correlacionadas

correlacion = np.corrcoef(
    ingresos_sim[:, 0],
    aprovisionamientos_sim[:, 0]
)

print(correlacion)
```

**EJEMPLO RESUELTO** Con los ingresos y los aprovisionamientos correlacionados, calcula el VAN del proyecto en cada simulación y compáralo con el obtenido suponiendo que son independientes (Pregunta 4). ¿Qué ocurre con el riesgo del proyecto?

```python
# Mismo cálculo que en la Pregunta 4, pero con los ingresos y aprovisionamientos correlacionados
FC_corr = (ingresos_sim - aprovisionamientos_sim - amortizacion - costes_generales) * (1 - tipo_impositivo) + amortizacion
VAN_corr = -I0 + (FC_corr * factores_descuento).sum(axis=1)

# Tabla comparativa (:>16 alinea el texto a la derecha en 16 caracteres)
print(f"{'':25s}{'Independientes':>16s}{'Correlacionados':>18s}")
print(f"{'VAN medio':25s}{VAN_sim.mean():>16,.0f}{VAN_corr.mean():>18,.0f}")
print(f"{'Desviación típica':25s}{VAN_sim.std():>16,.0f}{VAN_corr.std():>18,.0f}")
print(f"{'Probabilidad VAN > 0':25s}{np.mean(VAN_sim > 0):>16.2%}{np.mean(VAN_corr > 0):>18.2%}")

# alpha=0.5 hace los histogramas semitransparentes para ver los dos a la vez
plt.figure(figsize=(10, 4))
plt.hist(VAN_sim, bins=50, alpha=0.5, label="Independientes")
plt.hist(VAN_corr, bins=50, alpha=0.5, label="Correlacionados (ρ = 0.70)")
plt.axvline(0, color="red", linestyle="--")
plt.xlabel("VAN (€)")
plt.ylabel("Frecuencia")
plt.legend()
plt.grid(True)
plt.show()

# Cuando los ingresos suben, los aprovisionamientos también tienden a subir, de modo que
# parte de la variación se compensa: el VAN medio apenas cambia, pero su dispersión (riesgo) se reduce.
```
