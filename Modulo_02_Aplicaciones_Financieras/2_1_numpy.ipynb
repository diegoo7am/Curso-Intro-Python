**EJEMPLO RESUELTO** Disponemos de la siguiente información, estima el Beneficio Antes de Impuestos

**Fórmulas:**

$$BAI = \text{Ingresos} - \text{Aprovisionamiento} - \text{Amortización} - \text{Costes generales}$$

**¿Qué es la amortización?** Es el gasto contable que refleja el desgaste de los activos (maquinaria, instalaciones...). Reduce el beneficio, pero no supone una salida de dinero en ese año.

```python
# Todos los vectores tienen shape (5,): un valor por año
ingresos = np.array([            # shape (5,)
    300000,
    330000,
    350000,
    380000,
    400000
])

aprovisionamiento = np.array([   # shape (5,)
    120000,
    130000,
    138000,
    150000,
    158000
])

amortizacion = np.array([        # shape (5,)
    20000,
    22000,
    23000,
    24000,
    25000
])

costes_generales = np.array([    # shape (5,)
    60000,
    65000,
    68000,
    72000,
    75000
])

tipo_impositivo = 0.25
```

# Producto de matrices

**C = A @ B**

**C = np.dot(A, B)**

**Regla de las shapes:** las columnas de la primera deben coincidir con las filas de la segunda. El resultado se queda con las dimensiones de fuera:

$$(\text{filas}_A,\ \color{red}{n})\ @\ (\color{red}{n},\ \text{columnas}_B) \rightarrow (\text{filas}_A,\ \text{columnas}_B)$$

**EJEMPLO RESUELTO** Una tienda vende 3 tipos de bolígrafo. Determina el volumen de ventas.

|  | Producto 1 | Producto 2 | Producto 3 |
|---|---|---|---|
| **Precio (€)** | 3 | 12.5 | 28 |
| **Unidades vendidas** | 200 | 350 | 75 |

**Fórmulas:**

$$\text{Ventas} = \sum_{i} \text{precio}_i \times \text{unidades}_i$$

Esta suma de productos es justo lo que calcula `@` entre dos vectores.

```python
precios_boli = np.array([3, 12.5, 28])       # shape (3,): un precio por producto
unidades_tienda = np.array([200, 350, 75])   # shape (3,): unidades de cada producto

# (3,) @ (3,) -> un único número: multiplica cada precio por sus unidades y suma todo
ventas = precios_boli @ unidades_tienda      # equivalente a np.dot(precios_boli, unidades_tienda)
print(f"Volumen de ventas: {ventas:,.2f} €")   # :,.2f -> separador de miles y 2 decimales
```

**EJEMPLO RESUELTO** Un proyecto genera estos flujos de caja. Si la inversión es de 100.000 € y la rentabilidad exigida del 7 %, ¿conviene realizarlo?

|  | Año 1 | Año 2 | Año 3 | Año 4 | Año 5 |
|---|---|---|---|---|---|
| **Flujo de caja (€)** | 95.000 | 106.750 | 113.750 | 124.500 | 131.500 |

$$VAN=-I_0+\sum_{t=1}^T \frac{FC_t}{(1+k)^t}$$

$I_0$ es la inversión inicial: la escribimos en positivo y la restamos porque es un pago. $FC_t$ es el flujo de caja del año $t$ y $k$ la rentabilidad exigida. Si $VAN > 0$, la inversión crea valor y conviene realizarla.

```python
FC = np.array([95000, 106750, 113750, 124500, 131500])   # shape (5,): un flujo por año
t = np.arange(1, 6)                                      # shape (5,): años 1, 2, ..., 5
I0 = 100000                                              # inversión inicial
k = 0.07                                                 # rentabilidad exigida

# (5,) / (5,) -> (5,): cada flujo se divide entre (1 + k) elevado a su año
valores_actuales = FC / (1 + k) ** t                     # shape (5,)

# np.sum suma esos 5 valores actuales: el resultado es un único número
VAN = -I0 + np.sum(valores_actuales)
print(f"VAN = {VAN:,.2f} €")

# Si el VAN es positivo, la inversión crea valor
if VAN > 0:
    print("Conviene realizar la inversión")
else:
    print("No conviene realizar la inversión")
```

# Sistema de ecuaciones

## Determinante de una matriz
El determinante es un número asociado a una matriz cuadrada que indica, entre otras cosas, si la matriz puede invertirse.

La idea clave es:

Si $\det(A)\neq 0$ → la matriz tiene inversa y el sistema $Ax=b$ tiene solución única.

Si $\det(A)=0$ → la matriz no es invertible.

```python
A = np.array([   # shape (3, 3): tiene que ser cuadrada
    [2, 1, 0],
    [1, 3, 1],
    [0, 2, 4]
])
```

```python
np.linalg.det(A)
```

```python
inv_A = np.linalg.inv(A)   # shape (3, 3)
inv_A
```

```python
# (3, 3) @ (3, 3) -> (3, 3): la matriz identidad
I = inv_A @ A   # shape (3, 3)
I
```

# Broadcasting

Broadcasting permite que NumPy opere con arrays de dimensiones distintas cuando sus formas son compatibles.

NumPy “expande” virtualmente el array más pequeño para poder hacer la operación, sin necesidad de copiarlo manualmente.

## 1. Caso más sencillo: escalar + vector

Tengo unos precios de venta a los que tenemos que sumar una comisión constante de 5 €. Determinamos el precio final.

```python
precios = np.array([100, 120, 150, 200])   # shape (4,)

comision = 5

# (4,) + escalar -> (4,)
precios_finales = precios + comision       # shape (4,)

print(precios_finales)
```

Esto es lo que ha hecho Numpy

$$\begin{array}{lccl}
[100 & 120 & 150 & 200]\end{array}\\
+\\
\begin{array}{lccl}
[  5  & 5 &  5 &  5]
\end{array}$$

## 3. Aplicar un valor diferente a cada escenario

Ahora imaginemos que cada escenario tiene un factor distinto, concretamente queremos:

- Optimista → ×1.10
- Base → ×1.00
- Pesimista → ×0.90

```python
ventas = np.array([   # shape (3, 5): (escenarios, años)
    [320000, 355000, 390000, 425000, 460000],   # optimista
    [300000, 330000, 350000, 380000, 400000],   # base
    [280000, 300000, 315000, 330000, 345000]    # pesimista
])

factor = np.array([1.10, 1.00, 0.90])   # shape (3,): un factor por escenario
```

**La celda siguiente da error a propósito** (`ValueError`). Es lo esperado: queremos ver qué ocurre cuando las dimensiones no son compatibles.

```python
ventas_corregidas=ventas*factor
ventas_corregidas
```

Tenemos un problema, porque NumPy compara las dimensiones desde la derecha y aquí no coinciden (5 y 3):

- `ventas` → `(3, 5)`
- `factor` → `(3,)`

NumPy trataría `factor` como una fila repetida, con solo 3 columnas frente a las 5 de `ventas`:

$$\left[\begin{array}{lcr}
 1.10 & 1.00 & 0.90 \\
 1.10 & 1.00 & 0.90 \\
 1.10 & 1.00 & 0.90 \\
 \end{array}\right]$$

Tengo que convertir `factor` en una columna, de shape `(3, 1)`.

```python
factor = factor.reshape(3, 1)   # shape (3, 1)
factor
```

```python
factor.shape
```

```python
# (3, 5) * (3, 1) -> (3, 5): el factor de cada escenario se aplica a todos sus años
ventas_corregidas = ventas * factor   # shape (3, 5)
ventas_corregidas
```

## Regla para memorizar

NumPy compara las dimensiones desde la derecha. Dos dimensiones son compatibles si son iguales o si una de ellas vale 1.

| Primer array | Segundo array | ¿Compatible? | Se aplica |
|---|---|---|---|
| `(3, 5)` | `(5,)` | Sí | Un valor por año |
| `(3, 5)` | `(1, 5)` | Sí | Un valor por año |
| `(3, 5)` | `(3, 1)` | Sí | Un valor por escenario |
| `(3, 5)` | `(3,)` | **No** | Da error: 5 y 3 no coinciden |

## Sumar por filas o por columnas: `axis`

En una simulación tenemos una matriz con una fila por escenario y una columna por año. Para obtener un resultado por escenario hay que sumar cada fila.

- `axis=1`: suma cada fila y da un resultado por fila.
- `axis=0`: suma cada columna y da un resultado por columna.

```python
ejemplo = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(ejemplo.shape)         # (2, 3): 2 filas y 3 columnas
print(ejemplo.sum(axis=1))   # una suma por fila: 10+20+30 y 40+50+60
print(ejemplo.sum(axis=0))   # una suma por columna: 10+40, 20+50 y 30+60
```

**EJEMPLO RESUELTO** Tenemos que determinar cuánto pagaríamos por una empresa que no tiene deuda. El coste de capital es del 7 %. Para ello vamos a generar 10.000 posibles escenarios y estimar el valor de la empresa como una variable aleatoria.

Hipótesis:
- Los ingresos medios de los próximos 5 años son 500.000, 550.000, 600.000, 650.000 y 700.000 €, con una desviación típica del 10 %.
- El flujo de caja libre de cada año es el 20 % de los ingresos.
- A partir del año 5 el flujo de caja crece un 2 % anual a perpetuidad: $VT_5=\frac{FC_5\times(1+g)}{k-g}$

$$Valor=\sum_{t=1}^{5}\frac{FC_t}{(1+k)^t}+\frac{VT_5}{(1+k)^5}$$

**¿Qué es el valor terminal?** Es el valor, en el año 5, de todos los flujos de caja posteriores, suponiendo que crecen a un ritmo $g$ para siempre. Por eso hay que descontarlo también al presente.

```python
np.random.seed(42)   # fija los números aleatorios para que el resultado sea siempre el mismo

ingresos = np.array([500000, 550000, 600000, 650000, 700000])   # shape (5,)
k = 0.07                                                        # coste de capital
g = 0.02                                                        # crecimiento a perpetuidad
t = np.arange(1, 6)                                             # shape (5,): años 1 a 5

# Simulamos los ingresos: una fila por escenario y una columna por año
z = np.random.normal(0, 1, size=(10000, 5))                     # shape (10000, 5)
Ingresos_sim = ingresos + ingresos * 0.10 * z                   # shape (10000, 5)
FC_sim = Ingresos_sim * 0.20                                    # shape (10000, 5)

# (10000, 5) / (5,) -> (10000, 5). axis=1 suma por filas: un valor actual por escenario
VA_FC = (FC_sim / (1 + k) ** t).sum(axis=1)                     # shape (10000,)

# [:, -1] selecciona la última columna: el flujo del año 5 de cada escenario
VT = FC_sim[:, -1] * (1 + g) / (k - g)                          # shape (10000,)

valor_empresa = VA_FC + VT / (1 + k) ** 5                       # shape (10000,): un valor por escenario

valor_empresa.shape
```

```python
import matplotlib.pyplot as plt

# valor_empresa viene de la celda anterior: shape (10000,)
print(f"Valor medio de la empresa: {valor_empresa.mean():,.0f} €")
print(f"Desviación típica:         {valor_empresa.std():,.0f} €")
# Percentil 5 %: solo en el 5 % de los escenarios el valor es menor que esta cifra
print(f"Percentil 5 %:             {np.percentile(valor_empresa, 5):,.0f} €")
print(f"Percentil 95 %:            {np.percentile(valor_empresa, 95):,.0f} €")

plt.figure(figsize=(10, 4))
plt.hist(valor_empresa, bins=50)
plt.axvline(valor_empresa.mean(), color="black", linestyle="--", label="Media")   # línea vertical en la media
plt.xlabel("Valor de la empresa (€)")
plt.ylabel("Frecuencia")
plt.title("Distribución del valor de la empresa")
plt.legend()
plt.grid(True)
plt.show()
```

**EJEMPLO RESUELTO - Pregunta 4** Si la desviación típica de los ingresos representa el 15% de la media y la desviación típica del aprovisionamiento 8% de la media. Realiza un análisis de Monte Carlo con 10.000 simulaciones y analiza los resultados.

**Fórmulas:**

$$\text{Ingresos} = \mu_I + \sigma_I \times z, \quad \sigma_I = 0.15 \times \mu_I \qquad\qquad \text{Aprov.} = \mu_A + \sigma_A \times z, \quad \sigma_A = 0.08 \times \mu_A$$

con $z \sim N(0,1)$ distinto para cada variable. Para cada simulación se calcula el VAN y, al final:

$$P(VAN > 0) = \frac{\text{nº de simulaciones con } VAN > 0}{\text{nº total de simulaciones}}$$

```python
import matplotlib.pyplot as plt

# Datos del proyecto, repetidos para tenerlos a la vista
ingresos = np.array([500000, 550000, 600000, 650000, 700000])            # shape (5,)
aprovisionamientos = np.array([180000, 195000, 210000, 225000, 240000])  # shape (5,)
amortizacion = np.array([50000, 50000, 50000, 50000, 50000])             # shape (5,)
costes_generales = np.array([90000, 95000, 100000, 105000, 110000])      # shape (5,)
tipo_impositivo = 0.25
I0 = 1_000_000
k = 0.08
t = np.arange(1, 6)                                                      # shape (5,)
factores_descuento = 1 / (1 + k) ** t                                    # shape (5,)
n_simulaciones = 10000

np.random.seed(42)   # fija los números aleatorios para que el resultado sea siempre el mismo

# Shocks independientes N(0,1): una fila por simulación y una columna por año
z_ingresos = np.random.normal(0, 1, size=(n_simulaciones, 5))   # shape (10000, 5)
z_aprov = np.random.normal(0, 1, size=(n_simulaciones, 5))      # shape (10000, 5)

# Media + desviación típica × shock (15 % de la media para ingresos y 8 % para aprovisionamientos)
ingresos_sim = ingresos + ingresos * 0.15 * z_ingresos          # shape (10000, 5)
aprov_sim = aprovisionamientos + aprovisionamientos * 0.08 * z_aprov   # shape (10000, 5)

# Flujo de caja de cada simulación y año
FC_sim = (ingresos_sim - aprov_sim - amortizacion - costes_generales) * (1 - tipo_impositivo) + amortizacion   # shape (10000, 5)

# (10000, 5) * (5,) -> (10000, 5). axis=1 suma por filas: un VAN por simulación
VAN_sim = -I0 + (FC_sim * factores_descuento).sum(axis=1)       # shape (10000,)

print(f"VAN medio:            {VAN_sim.mean():,.0f} €")
print(f"Desviación típica:    {VAN_sim.std():,.0f} €")
# VAN_sim > 0 da True/False en cada simulación. True cuenta como 1 y False como 0,
# así que la media de esos valores es la proporción de simulaciones con VAN positivo
print(f"Probabilidad VAN > 0: {np.mean(VAN_sim > 0):.2%}")
# Percentil 5 %: solo en el 5 % de las simulaciones el VAN es menor que esta cifra
print(f"Percentil 5 %:        {np.percentile(VAN_sim, 5):,.0f} €")
print(f"Percentil 95 %:       {np.percentile(VAN_sim, 95):,.0f} €")

plt.figure(figsize=(10, 4))
plt.hist(VAN_sim, bins=50)
plt.axvline(0, color="red", linestyle="--", label="VAN = 0")
plt.xlabel("VAN (€)")
plt.ylabel("Frecuencia")
plt.title("Distribución del VAN (10.000 simulaciones)")
plt.legend()
plt.grid(True)
plt.show()
```

**EJEMPLO RESUELTO** Suponemos que los ingresos y los aprovisionamientos tienen un coeficiente de correlación $\rho=0.70$. Calcula el VAN del proyecto en cada simulación y compáralo con el que se obtiene si son independientes. ¿Qué ocurre con el riesgo del proyecto?

Para introducir la correlación, construimos el shock del aprovisionamiento a partir del shock de los ingresos:

$$z_A=\rho Z_I+\sqrt{1-\rho^2}Z_2$$

donde:
- $Z_I\sim N(0,1)$ shock de los ingresos
- $Z_2\sim N(0,1)$ shock independiente del aprovisionamiento
- $\rho$ correlación
- $z_A$ el shock del aprovisionamiento

Después transformamos cada shock en su variable: $\text{Ingresos}=\mu_I + \sigma_I z_I$ y $\text{Aprov.}=\mu_A + \sigma_A z_A$.

```python
# Datos del proyecto, repetidos para tenerlos a la vista
ingresos = np.array([500000, 550000, 600000, 650000, 700000])            # shape (5,)
aprovisionamientos = np.array([180000, 195000, 210000, 225000, 240000])  # shape (5,)
amortizacion = np.array([50000, 50000, 50000, 50000, 50000])             # shape (5,)
costes_generales = np.array([90000, 95000, 100000, 105000, 110000])      # shape (5,)
tipo_impositivo = 0.25
I0 = 1_000_000
k = 0.08
t = np.arange(1, 6)                                                      # shape (5,)
factores_descuento = 1 / (1 + k) ** t                                    # shape (5,)
n_simulaciones = 10000
rho = 0.70

np.random.seed(42)   # fija los números aleatorios para que el resultado sea siempre el mismo

# Dos matrices de shocks N(0,1) independientes: una fila por simulación y una columna por año
z_ingresos = np.random.normal(0, 1, size=(n_simulaciones, 5))   # shape (10000, 5)
z_2 = np.random.normal(0, 1, size=(n_simulaciones, 5))          # shape (10000, 5)

# Shock del aprovisionamiento correlacionado con el de los ingresos
z_A = rho * z_ingresos + np.sqrt(1 - rho**2) * z_2              # shape (10000, 5)

# Los ingresos simulados son los mismos en los dos casos
ingresos_sim = ingresos + ingresos * 0.15 * z_ingresos          # shape (10000, 5)
aprov_indep = aprovisionamientos + aprovisionamientos * 0.08 * z_2   # shape (10000, 5)
aprov_corr = aprovisionamientos + aprovisionamientos * 0.08 * z_A    # shape (10000, 5)

# Flujos de caja: uno con aprovisionamientos independientes y otro con correlacionados
FC_indep = (ingresos_sim - aprov_indep - amortizacion - costes_generales) * (1 - tipo_impositivo) + amortizacion   # shape (10000, 5)
FC_corr = (ingresos_sim - aprov_corr - amortizacion - costes_generales) * (1 - tipo_impositivo) + amortizacion     # shape (10000, 5)

# axis=1 suma por filas: un VAN por simulación
VAN_indep = -I0 + (FC_indep * factores_descuento).sum(axis=1)   # shape (10000,)
VAN_corr = -I0 + (FC_corr * factores_descuento).sum(axis=1)     # shape (10000,)

# Comprobamos la correlación en el primer año (columna 0): debe salir cerca de 0.70
print(np.corrcoef(ingresos_sim[:, 0], aprov_corr[:, 0]))
```
