## Determinante de una matriz
El determinante es un número asociado a una matriz cuadrada que indica, entre otras cosas, si la matriz puede invertirse.

La idea clave para tus alumnos es:

Si $\det(A)\neq 0$ → la matriz tiene inversa y el sistema \(Ax=b\) tiene solución única.

Si $\det(A)=0$ → la matriz no es invertible.

```python
A = np.array([
    [2, 1, 0],
    [1, 3, 1],
    [0, 2, 4]
])
```

```python
np.linalg.det(A)
```

```python
inv_A=np.linalg.inv(A)
inv_A
```

```python
I=inv_A @ A
I
```

# Broadcasting

Broadcasting permite que NumPy opere con arrays de dimensiones distintas cuando sus formas son compatibles.

NumPy “expande” virtualmente el array más pequeño para poder hacer la operación, sin necesidad de copiarlo manualmente.

## 1. Caso más sencillo: escalar + vector

Tengo unos precios de ventas a los que que tenemos que suma una comisión constante de 5 €. Determinamos el precio final.

```python
precios = np.array([100, 120, 150, 200])

comision = 5

precios_finales = precios + comision

print(precios_finales)
```

Esto es lo que ha hecho Numpy

$$\begin{array}{lccl}
[100 & 120 & 150 & 200]\end{array}\\
+\\
\begin{array}{lccl}
[  5  & 5 &  5 &  5]
\end{array}$$

**EJERCICIO** Suponemos que los ingresos y los costes tienen un coeficiente de correlación  $\rho=0.70$

```python
ingresos = np.array([
    500000,
    550000,
    600000,
    650000,
    700000
])

aprovisionamientos = np.array([
    180000,
    195000,
    210000,
    225000,
    240000
])

desv_ingresos = ingresos * 0.15
desv_aprovisionamientos = aprovisionamientos * 0.08

rho = 0.70

n_simulaciones = 10000
years = np.arange(1, 6)   # 5 años de vida del proyecto
```

Vamos a construir dos matrices aleatorias $N(0,1)$ independientes: $Z_1$ y $Z_2$

```python
z1 = np.random.normal(
    0,
    1,
    size=(n_simulaciones, len(years))
)

z2 = np.random.normal(
    0,
    1,
    size=(n_simulaciones, len(years))
)
```

Introducimos la correlación la función de distribución estandarizada del aprovisionamiento será:

$$z_A=\rho Z_I+\sqrt{1-\rho^2}Z_2$$

donde:
- $Z_I\sim N(0,1)$ shock de los ingresos
- $Z_2\sim N(0,1)$ shock indepediente del aprovisionamiento
- $\rho$ correlación
- $z_A$ el schock del aprovisionamiento

```python
z_A = (
    rho * z1
    + np.sqrt(1 - rho**2) * z2
)
```

Transformamos el shock $z_I$ en ingresos

$$I=\mu_I + \sigma z_I$$

```python
ingresos_sim = (
    ingresos
    + desv_ingresos * z1
)
ingresos_sim
```

Transformamos el shock $z_A$ en costes de aprovisionamiento

$$Aprov=\mu_A + \sigma z_A$$

```python
aprovisionamientos_sim = (
    aprovisionamientos
    + desv_aprovisionamientos * z_A
)
aprovisionamientos_sim
```

```python
# Comprobamos si están correlacionadas

correlacion = np.corrcoef(
    ingresos_sim[:, 0],
    aprovisionamientos_sim[:, 0]
)

print(correlacion)
```
